```python
import numpy as np
def newtonian_velocity(r,M):
    G=4.300e-3
    v=np.sqrt(G*M/r)
    return v
galaxy_radii=np.array([1.0,2.0,5.0,10.0])
central_mass=1e10
predicted_velocities= newtonian_velocity(galaxy_radii,central_mass)
print("Predicted velocities", predicted_velocities)
```

```python
import numpy as np
import matplotlib.pyplot as plt
r_smooth=np.linspace(0.5,15, 100)
G=4.300e-3
M=1e10
v_smooth=np.sqrt(G*M/r_smooth)
plt.figure(figsize=(8,5))
plt.plot(r_smooth, v_smooth, label='Newtonian Prediction', color='red', linewidth=2)
plt.title("Galaxy Rotation Curve Newtonian Model")
plt.xlabel("Radius (kpc)")
plt.ylabel("Velocity (km/s)")
# set legend appearance and show single legend
plt.legend(frameon=True, framealpha=0.6)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
r=np.linspace(0.1,20,200)
rs=5.0
nfw_mass_factor=np.log(1+rs/r)-(r/rs)
v_dark_matter=150*np.sqrt(nfw_mass_factor/r)
plt.figure(figsize=(8,5))
plt.plot(r, v_dark_matter, label='Dark Matter Velocity', color='blue',linewidth=2)
title="Velocity Contribution from Dark Matter Halo (NFW Profile)"
xlabel="Radius (kpc)"
ylabel="Velocity (km/s)"
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
